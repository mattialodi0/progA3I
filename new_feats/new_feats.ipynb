```python
import pandas as pd
```

```python
# make a single dataframe with all the new features
df_climate = pd.read_csv('../datasets/disasters_new_feats/climate.csv')
df_elevation = pd.read_csv('../datasets/disasters_new_feats/elevation.csv')
df_land_coverage = pd.read_csv('../datasets/disasters_new_feats/land_coverage.csv')
df_river_distance = pd.read_csv('../datasets/disasters_new_feats/river_distance.csv')
df_sea_distance = pd.read_csv('../datasets/disasters_new_feats/sea_distance.csv')
```

```python
df_land_coverage.drop(columns=['BEGIN_LAT', 'BEGIN_LON'], inplace=True)
df_river_distance.drop(columns=['BEGIN_LAT', 'BEGIN_LON'], inplace=True)
df_sea_distance.drop(columns=['BEGIN_LAT', 'BEGIN_LON'], inplace=True)
```

```python
df = df_climate
df = df.merge(df_elevation, right_index=True, left_index=True)
df = df.merge(df_land_coverage, right_index=True, left_index=True)
df = df.merge(df_river_distance, right_index=True, left_index=True)
df = df.merge(df_sea_distance, right_index=True, left_index=True)
```

```python
# rename cols
df.rename(columns={
    'ANNUAL_PRECIPITATION': 'PRECIPITATION',
    'BARREN': 'COV_BARREN',
    'CULTIVATED': 'COV_CULTIVATED',
    'VEGETATION': 'COV_VEGETATION',
    'FOREST': 'COV_FOREST',
    'WATER': 'COV_WATER',
    'SNOW_ICE': 'COV_SNOW_ICE',
    'URBAN': 'COV_URBAN'
}, inplace=True)

df.head()
```

```python
df.to_csv('../datasets/disasters_new_feats/all_new_feats.csv', index=False)
```
